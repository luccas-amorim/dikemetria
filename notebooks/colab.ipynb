````python
from google.colab import drive

drive.mount("/content/drive")
````

````python
from dikemetria.analise import analisar_documento
from dikemetria.extracao import extrair_texto
from dikemetria.pseudonimizacao import pseudonimizar

# Atualize o caminho: PDF, DOCX, TXT ou HTML
caminho = "/content/drive/My Drive/JI/data/decisao.docx"

texto = pseudonimizar(extrair_texto(caminho)).texto
analise = analisar_documento(texto)
print("Resultado:", analise.resultado, "|", analise.evidencia_resultado)
print("Citações:", analise.citacoes)
print("Termos:", analise.termos_juridicos)
````

````python
for argumento in analise.argumentos:
    print("-", argumento)
````
